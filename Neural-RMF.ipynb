# NEURAL-RMF — Pipeline Demo

This notebook demonstrates the full NEURAL-RMF EEG monitoring pipeline on a real EDF recording.

The system:
1. Loads an EDF file (CHB-MIT or Siena format)
2. Calibrates on the first 8 minutes of recording
3. Runs rolling inference and outputs a 3-state alert semaphore per window

**No seizure annotations required** — the pipeline works on any EEG recording.

---

### Alert states

| State | Meaning |
|-------|--------|
| `CONOCIDO` | Baseline interictal activity |
| `FRONTERA` | Approaching elevated territory — increased vigilance |
| `OPUESTO` | Clear pre-ictal signal — alert |

## 2. Imports and configuration

```python
import sys
import os
import numpy as np

# Add the package root to the path if running from inside the package directory
sys.path.insert(0, os.path.abspath(''))

from core.eeg_encoder import EEGEncoder
from core.field_engine import build_field, measure_window
from core.alert_system import AlertSystem
```

```python
# ── Configuration ─────────────────────────────────────────────────────────────

EDF_PATH     = "data/example_eeg.edf"   # path to your EDF file
EEG_CHANNELS = ["F7", "T7", "F8", "T8"] # frontal-temporal 4-channel montage
FS           = 256                       # sampling rate in Hz (256 for CHB-MIT, 512 for Siena)
CALIB_MIN    = 8                         # calibration duration in minutes
WIN_SEC      = 2.0                       # analysis window length in seconds

# Field and alert parameters (validated on CHB-MIT and Siena)
CALIB_OFFSET = [1.5, 0.0, 0.0]
OMEGA_SCALE  = 1.5
THR_COL_ABS  = 0.70

CALIB_WINS = int(CALIB_MIN * 60 / WIN_SEC)
print(f"Calibration windows: {CALIB_WINS} ({CALIB_MIN} min at {WIN_SEC}s each)")
```

## 3. Load EDF file

```python
import pyedflib

def load_edf(path, channels, fs_expected):
    """Return (data, fs) where data is ndarray (n_ch, n_samples)."""
    f = pyedflib.EdfReader(path)
    labels = [l.strip() for l in f.getSignalLabels()]
    fs = int(f.getSampleFrequency(0))
    idx = [i for i, l in enumerate(labels) if l in channels]
    if not idx:
        raise ValueError(f"None of {channels} found in {path}.\nAvailable: {labels}")
    data = np.array([f.readSignal(i) for i in idx])
    f._close()
    print(f"Loaded: {path}")
    print(f"  Channels found : {[labels[i] for i in idx]}")
    print(f"  Sampling rate  : {fs} Hz")
    print(f"  Duration       : {data.shape[1]/fs/60:.1f} min ({data.shape[1]} samples)")
    return data, fs

data, fs = load_edf(EDF_PATH, EEG_CHANNELS, FS)
```

## 4. Calibration phase

The first `CALIB_MIN` minutes are used to learn the patient's individual interictal baseline.
No labels or annotations are needed.

```python
win_samp = int(WIN_SEC * fs)

# Extract calibration windows
calib_windows = []
start = 0
while start + win_samp <= data.shape[1] and len(calib_windows) < CALIB_WINS:
    calib_windows.append(data[:, start:start + win_samp])
    start += win_samp

print(f"Calibration: {len(calib_windows)} windows collected")

# Fit the EEG encoder on calibration data
enc = EEGEncoder(
    fs=fs,
    win_sec=WIN_SEC,
    calib_offset=CALIB_OFFSET,
    omega_scale=OMEGA_SCALE,
)
enc.fit(calib_windows)
print("EEGEncoder fitted.")

# Build the internal field (opaque binary module)
field = build_field(seed=42)

# Warm up the field with calibration windows and collect calibration novelties
calib_novelties = []
for win in calib_windows:
    omega = enc.encode(win)
    m = measure_window(field, omega)
    calib_novelties.append(m["nov_collective"])

# Build the alert system (sets FRONTERA threshold at P80 of calibration)
alert = AlertSystem(
    thr_col_abs=THR_COL_ABS,
    calib_novelties=calib_novelties,
)
alert.reset()

print(f"\nCalibration complete.")
print(f"  FRONTERA threshold (P80) : {alert.thr_amarillo:.3f}")
print(f"  OPUESTO  threshold (abs) : {THR_COL_ABS:.2f}")
```

## 5. Rolling inference

Each 2-second window is encoded into a 3D vector, passed through the field engine,
and evaluated by the alert system. Results are collected minute by minute.

```python
results = []
win_idx = 0

# Start after calibration segment
pos = start

while pos + win_samp <= data.shape[1]:
    win = data[:, pos:pos + win_samp]
    pos += win_samp
    win_idx += 1

    omega   = enc.encode(win)
    metrics = measure_window(field, omega)
    state   = alert.update(metrics)

    t_min = round(win_idx * WIN_SEC / 60.0, 2)

    results.append({
        "t_min"        : t_min,
        "nov_max"      : round(state.nov_max, 4),
        "nov_collective": round(state.nov_collective, 4),
        "state"        : state.state,
        "alert_level"  : state.alert_level,
        "lead_minutes" : state.lead_minutes,
    })

print(f"Inference complete: {len(results)} windows processed.")
```

## 6. Results

```python
# Print a compact table of results
print(f"{'t_min':>7}  {'nov_max':>8}  {'nov_col':>8}  {'state':>10}  {'alert':>9}  {'lead':>7}")
print("-" * 65)

for r in results:
    marker = ""
    if r["alert_level"] == "warn":
        marker = "  [FRONTERA]"
    elif r["alert_level"] == "critical":
        marker = "  [OPUESTO - ALERT]"

    print(
        f"{r['t_min']:>7.1f}  "
        f"{r['nov_max']:>8.3f}  "
        f"{r['nov_collective']:>8.3f}  "
        f"{r['state']:>10}  "
        f"{r['alert_level']:>9}  "
        f"{str(r['lead_minutes']):>7}"
        f"{marker}"
    )
```

```python
# Summary statistics
from collections import Counter

state_counts = Counter(r["state"] for r in results)
alert_counts = Counter(r["alert_level"] for r in results)

print("State distribution:")
for state, count in state_counts.most_common():
    pct = 100 * count / len(results)
    print(f"  {state:<12} {count:>4} windows  ({pct:.1f}%)")

print("\nAlert level distribution:")
for level, count in alert_counts.most_common():
    pct = 100 * count / len(results)
    print(f"  {level:<10} {count:>4} windows  ({pct:.1f}%)")

# First alert time
warn_times = [r["t_min"] for r in results if r["alert_level"] == "warn"]
crit_times = [r["t_min"] for r in results if r["alert_level"] == "critical"]

if warn_times:
    print(f"\nFirst FRONTERA alert : {min(warn_times):.1f} min")
if crit_times:
    print(f"First OPUESTO alert  : {min(crit_times):.1f} min")
if results and results[-1]["lead_minutes"] is not None:
    print(f"Lead time at end     : {results[-1]['lead_minutes']:.1f} min")
```

## Notes

**Channel mapping**
- CHB-MIT recordings: channels are labeled `F7`, `T7`, `F8`, `T8` at 256 Hz.
- Siena recordings: same channel names, 512 Hz — update `FS = 512` in the configuration cell.
- If your EDF uses different channel labels, update `EEG_CHANNELS` to match exactly.

**Calibration window**
- The first `CALIB_MIN` minutes must be interictal (seizure-free) baseline.
- If the recording starts during or close to a seizure, increase `CALIB_MIN` accordingly.

**Alert thresholds**
- `FRONTERA` (orange) threshold is set automatically at the P80 of calibration novelty — it adapts to the individual.
- `OPUESTO` (red) threshold (`THR_COL_ABS = 0.70`) is fixed and validated across both datasets.
- An alert is confirmed only after `confirm_windows = 3` consecutive elevated windows (~1 min).

**Core engine**
- `core/field_engine` is distributed as a compiled binary (`.pyd` on Windows, `.so` on Linux/macOS).
- `core/eeg_encoder.py` and `core/alert_system.py` are provided as readable Python source.
